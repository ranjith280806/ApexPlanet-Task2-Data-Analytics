```python
# ============================================================
# APEXPLANET INTERNSHIP - TASK 2
# EXPLORATORY DATA ANALYSIS (EDA) & BUSINESS INTELLIGENCE
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

sns.set_theme(style="whitegrid")

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 120)

os.makedirs("/content/outputs", exist_ok=True)


# ============================================================
# 2. LOAD CLEANED DATASET
# ============================================================

data = pd.read_excel("/content/Cleaned_Dataset.xlsx")

print("Dataset loaded successfully!")

print("\nFirst 5 Rows")
print(data.head())


# ============================================================
# 3. BASIC DATASET INFORMATION
# ============================================================

print("\n" + "="*60)
print("DATASET INFORMATION")
print("="*60)

print("\nShape of Dataset:")
print(data.shape)

print("\nNumber of Rows:", data.shape[0])
print("Number of Columns:", data.shape[1])

print("\nColumn Names:")
print(data.columns.tolist())

print("\nData Types:")
print(data.dtypes)

print("\nMissing Values:")
print(data.isnull().sum())

print("\nTotal Missing Values:")
print(data.isnull().sum().sum())

print("\nDuplicate Rows:")
print(data.duplicated().sum())


# ============================================================
# 4. DATE CONVERSION
# ============================================================

data["Order_Date"] = pd.to_datetime(
    data["Order_Date"],
    errors="coerce"
)

print("\nOrder_Date converted to datetime.")


# ============================================================
# 5. DESCRIPTIVE STATISTICS
# ============================================================

print("\n" + "="*60)
print("DESCRIPTIVE STATISTICS")
print("="*60)

print("\nNumerical Summary:")

print(
    data[
        [
            "Age",
            "Quantity",
            "Unit_Price",
            "Total_Sales"
        ]
    ].describe()
)


# ============================================================
# 6. CATEGORICAL ANALYSIS
# ============================================================

print("\n" + "="*60)
print("CATEGORICAL ANALYSIS")
print("="*60)

print("\nGender Distribution:")
print(data["Gender"].value_counts())

print("\nCategory Distribution:")
print(data["Category"].value_counts())

print("\nCity Distribution:")
print(data["City"].value_counts())

print("\nProduct Distribution:")
print(data["Product"].value_counts())


# ============================================================
# 7. UNIVARIATE ANALYSIS
# ============================================================


# ------------------------------------------------------------
# 7.1 AGE DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

sns.histplot(
    data["Age"],
    bins=15,
    kde=True
)

plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Customers")

plt.tight_layout()

plt.savefig(
    "/content/outputs/age_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# 7.2 SALES DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

sns.histplot(
    data["Total_Sales"],
    bins=30,
    kde=True
)

plt.title("Total Sales Distribution")
plt.xlabel("Total Sales")
plt.ylabel("Frequency")

plt.tight_layout()

plt.savefig(
    "/content/outputs/sales_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# 7.3 SALES BY CATEGORY
# ------------------------------------------------------------

category_sales = (
    data.groupby("Category")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("\nSales by Category:")
print(category_sales)


plt.figure(figsize=(10, 6))

sns.barplot(
    x=category_sales.index,
    y=category_sales.values
)

plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.xticks(rotation=30)

plt.tight_layout()

plt.savefig(
    "/content/outputs/sales_by_category.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# 7.4 SALES BY CITY
# ------------------------------------------------------------

city_sales = (
    data.groupby("City")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("\nSales by City:")
print(city_sales)


plt.figure(figsize=(10, 6))

sns.barplot(
    x=city_sales.index,
    y=city_sales.values
)

plt.title("Total Sales by City")
plt.xlabel("City")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    "/content/outputs/sales_by_city.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 8. MONTHLY SALES TREND
# ============================================================

data["Month"] = data["Order_Date"].dt.to_period("M")

monthly_sales = (
    data.groupby("Month")["Total_Sales"]
    .sum()
)

print("\nMonthly Sales:")
print(monthly_sales)


plt.figure(figsize=(12, 6))

plt.plot(
    monthly_sales.index.astype(str),
    monthly_sales.values,
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    "/content/outputs/monthly_sales.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 9. TOP 10 PRODUCTS BY SALES
# ============================================================

top_products = (
    data.groupby("Product")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print("\nTop 10 Products by Revenue:")
print(top_products)


plt.figure(figsize=(12, 6))

sns.barplot(
    x=top_products.values,
    y=top_products.index
)

plt.title("Top 10 Products by Revenue")
plt.xlabel("Total Sales")
plt.ylabel("Product")

plt.tight_layout()

plt.savefig(
    "/content/outputs/top_products.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 10. SALES BY GENDER
# ============================================================

gender_sales = (
    data.groupby("Gender")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("\nSales by Gender:")
print(gender_sales)


plt.figure(figsize=(8, 5))

sns.barplot(
    x=gender_sales.index,
    y=gender_sales.values
)

plt.title("Total Sales by Gender")
plt.xlabel("Gender")
plt.ylabel("Total Sales")

plt.tight_layout()

plt.savefig(
    "/content/outputs/sales_by_gender.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 11. MULTIVARIATE ANALYSIS
# ============================================================


# ------------------------------------------------------------
# 11.1 QUANTITY VS TOTAL SALES
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=data,
    x="Quantity",
    y="Total_Sales",
    hue="Category"
)

plt.title("Quantity vs Total Sales")
plt.xlabel("Quantity")
plt.ylabel("Total Sales")

plt.tight_layout()

plt.savefig(
    "/content/outputs/sales_vs_quantity.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 12. AGE VS TOTAL SALES
# ============================================================

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=data,
    x="Age",
    y="Total_Sales",
    hue="Gender"
)

plt.title("Age vs Total Sales")
plt.xlabel("Age")
plt.ylabel("Total Sales")

plt.tight_layout()

plt.savefig(
    "/content/outputs/age_vs_sales.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 13. CATEGORY VS GENDER
# ============================================================

plt.figure(figsize=(10, 6))

sns.countplot(
    data=data,
    x="Category",
    hue="Gender"
)

plt.title("Category Distribution by Gender")
plt.xlabel("Category")
plt.ylabel("Number of Orders")

plt.xticks(rotation=30)

plt.tight_layout()

plt.savefig(
    "/content/outputs/category_vs_gender.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 14. CORRELATION ANALYSIS
# ============================================================

numeric_columns = [
    "Age",
    "Quantity",
    "Unit_Price",
    "Total_Sales"
]

correlation = data[numeric_columns].corr()

print("\n" + "="*60)
print("CORRELATION MATRIX")
print("="*60)

print(correlation)


# ============================================================
# 15. CORRELATION HEATMAP
# ============================================================

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.tight_layout()

plt.savefig(
    "/content/outputs/correlation_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 16. KEY BUSINESS METRICS
# ============================================================

total_sales = data["Total_Sales"].sum()

total_orders = data["Order_ID"].nunique()

total_customers = data["Customer_ID"].nunique()

total_quantity = data["Quantity"].sum()

average_order_value = data["Total_Sales"].mean()

highest_order = data["Total_Sales"].max()


print("\n" + "="*60)
print("KEY BUSINESS METRICS")
print("="*60)

print(f"\nTotal Sales       : {total_sales:,.2f}")
print(f"Total Orders      : {total_orders}")
print(f"Total Customers   : {total_customers}")
print(f"Total Quantity    : {total_quantity}")
print(f"Average Order     : {average_order_value:,.2f}")
print(f"Highest Order     : {highest_order:,.2f}")


# ============================================================
# 17. BUSINESS INSIGHTS
# ============================================================

highest_category = category_sales.idxmax()
highest_category_sales = category_sales.max()

highest_city = city_sales.idxmax()
highest_city_sales = city_sales.max()

best_product = top_products.idxmax()
best_product_sales = top_products.max()

best_month = monthly_sales.idxmax()
best_month_sales = monthly_sales.max()


print("\n" + "="*60)
print("KEY BUSINESS INSIGHTS")
print("="*60)

print(
    f"\n1. The highest-selling category is "
    f"{highest_category}, generating "
    f"{highest_category_sales:,.2f} in sales."
)

print(
    f"\n2. The city with the highest sales is "
    f"{highest_city}, generating "
    f"{highest_city_sales:,.2f} in sales."
)

print(
    f"\n3. The top product by revenue is "
    f"{best_product}, generating "
    f"{best_product_sales:,.2f}."
)

print(
    f"\n4. The highest-sales month is "
    f"{best_month}, with sales of "
    f"{best_month_sales:,.2f}."
)

print(
    f"\n5. The average order value is "
    f"{average_order_value:,.2f}."
)


# ============================================================
# 18. SAVE FINAL ANALYSIS DATA
# ============================================================

data.to_excel(
    "/content/outputs/Task2_Analyzed_Dataset.xlsx",
    index=False
)

print("\nAnalyzed dataset saved successfully!")

print("\n" + "="*60)
print("TASK 2 EDA COMPLETED SUCCESSFULLY!")
print("="*60)


# ============================================================
# 19. DISPLAY OUTPUT FILES
# ============================================================

print("\nOutput files created:")

for file in os.listdir("/content/outputs"):
    print(file)
```
